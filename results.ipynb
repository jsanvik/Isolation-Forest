```python
import random
import numpy as np
import pandas as pd
import math
import statistics
from pathlib import Path
import os
import time
import matplotlib.pyplot as plt
```

```python
directory = Path(os.getcwd())
results = pd.read_csv(directory / "data" / f"results.csv")
```

```python
hbk = results[results["Dataset"] == "hbk"]
wood = results[results["Dataset"] == "wood"]
```

```python
# How do hlim and ntrees impact AUC?
plt.figure()
plt.scatter(wood["hlim"], wood["AUC"], alpha=0.7)
plt.xlabel("Height Limit"); plt.ylabel("AUC")

plt.figure()
plt.scatter(wood["ntrees"], wood["AUC"], alpha=0.7)
plt.xlabel("Number of Trees"); plt.ylabel("AUC")
```

```python
# How do hlim and ntrees impact Processing Time?
plt.figure()
plt.scatter(wood["hlim"], wood["Total Processing Time"], alpha=0.7)
plt.xlabel("Height Limit"); plt.ylabel("Total Processing Time")

plt.figure()
plt.scatter(wood["ntrees"], wood["Total Processing Time"], alpha=0.7)
plt.xlabel("Number of Trees"); plt.ylabel("Total Processing Time")
```
